# **Gradient Boosting  for classification**

## **Introduction**



**Gradient Boosting** is an ensemble method that builds a strong model by adding weak learners one at a time. The image above shows the main idea. The model starts with a simple prediction, computes the current errors, trains a weak learner to predict those errors, and then updates the model. This process repeats for many rounds.

In gradient boosting, the errors are often described as **residuals** or **negative gradients** of the loss function. Each new learner moves the model in a direction that reduces the loss. The learning rate controls how much each new learner changes the current model. A smaller learning rate usually makes training slower but can improve generalization.


### **AdaBoost vs Gradient Boost**

AdaBoost and Gradient Boosting are both ensemble methods that build a strong model by adding weak learners sequentially. The main difference is how each method decides what the next learner should focus on.

AdaBoost adjusts sample weights after each round. Observations that are misclassified receive larger weights, so the next learner pays more attention to those difficult cases.

Gradient Boosting takes a more general optimization approach. Each new learner is trained to reduce the current model's error by following the gradient of a chosen loss function. For tree-based gradient boosting, this usually means fitting each new tree to the residual errors left by the previous trees.

Conceptually, AdaBoost can be viewed as a special case of gradient boosting with exponential loss. In practice, scikit-learn provides them as separate estimators: `AdaBoostClassifier` and `GradientBoostingClassifier`.

Because Gradient Boosting adds learners one at a time, it can overfit if the ensemble becomes too complex. Important controls include `n_estimators`, `learning_rate`, tree depth, subsampling, regularization, and early stopping.



The comparison image above summarizes the difference. AdaBoost pays more attention to misclassified samples by changing sample weights. Gradient Boosting fits each new learner to the residual errors, then adds that learner to the current model using a learning rate.

In this lab, we will focus on tree-based boosters for classification.

XGBoost follows the gradient boosting idea shown above, but adds practical improvements such as regularization, row and column sampling, efficient tree building, and strong hyperparameter controls. These tools often make XGBoost accurate and efficient on tabular datasets.

### **Defining Helper Functions**

This function will calculate the accuracy of the training and testing data given a model.


```python
def get_accuracy(X_train, X_test, y_train, y_test, model):
    return  {"test Accuracy":metrics.accuracy_score(y_test, model.predict(X_test)),"train Accuracy": metrics.accuracy_score(y_train, model.predict(X_train))}
```

This function calculates the average accuracy of differnt learning rates on training and test data.


```python
def get_accuracy_boost(X,y,title,times=20,xlabel='Number Estimators',Learning_rate_=[0.2,0.4,0.6,1], n_est = 100):

    lines_array=['solid','--', '-.', ':']

    N_estimators=[n*2 for n in range(1,n_est//2)]
    
    train_acc=np.zeros((times,len(Learning_rate_),len(N_estimators)))
    test_acc=np.zeros((times,len(Learning_rate_),len(N_estimators)))


    #Iterate through different number of Learning rate  and average out the results  
    
    for n in tqdm(range(times)):
        X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.3, stratify=y)
        for n_estimators in N_estimators:
            for j,lr in enumerate(Learning_rate_):


                model = XGBClassifier(objective=objective,learning_rate=lr,n_estimators=n_estimators,eval_metric='mlogloss')


                model.fit(X_train,y_train)



                Accuracy=get_accuracy(X_train, X_test, y_train, y_test,  model)



                train_acc[n,j,(n_estimators//2)-1]=Accuracy['train Accuracy']
                test_acc[n,j,(n_estimators//2)-1]=Accuracy['test Accuracy']
    



    fig, ax1 = plt.subplots()
    mean_test=test_acc.mean(axis=0)
    mean_train=train_acc.mean(axis=0)
    ax2 = ax1.twinx()

    for j,(lr,line) in enumerate(zip(Learning_rate_,lines_array)): 

        ax1.plot(mean_train[j,:],linestyle = line,color='b',label="Learning rate "+str(lr))
        ax2.plot(mean_test[j,:],linestyle = line, color='r',label=str(lr))

    ax1.set_ylabel('Training accuracy',color='b')
    ax1.legend()
    ax2.set_ylabel('Testing accuracy', color='r')
    ax2.legend()
    ax1.set_xlabel(xlabel)
    plt.show()
```

`mlogloss` cross entropy loss

`error` : number of misclassified samples

`auc` : area under  ROC curve (receiver operating characteristic curve)

We will use `error`:


```python
model = XGBClassifier(eval_metric='error') 
```

The validation data must be a list `eval_set`: the list  contains a tuple, the first element of the tuple must be the features, the second must be the labels:


```python
eval_set = [(X_test, y_test)]
```

```python
model.fit(X_train, y_train, eval_set=eval_set, verbose=True)
```

```python
plt.plot(range(0, len(results['validation_0']['error'])), results['validation_0']['error'])
plt.xlabel('iterations')
plt.ylabel('Misclassified Samples')
```

## **How  Gradient Boosting  Works (Optional)**


Gradient Boosting builds one strong model by adding many weak models one at a time. Like any supervised learning problem, we start with a dataset $\{(\mathbf{x}_1, y_1), \ldots, (\mathbf{x}_N, y_N)\}$. The final model $H_T(\mathbf{x})$ is a linear combination of $T$ weak models $h_t(\mathbf{x})$, which are usually small trees. The value $\alpha_t$ is the weight for weak model $t$. In many cases, $\alpha_t = 1$.

The key idea is that each new weak model focuses on the mistakes made by the models before it. So $h_t(\mathbf{x})$ is a separate model, but it is trained using information from $h_1(\mathbf{x}), \ldots, h_{t-1}(\mathbf{x})$.

$H_T(\mathbf{x}) = \sum_{t=1}^T \alpha_t h_t(\mathbf{x})$

Borrowing notation from <a href="https://www.cs.cornell.edu/courses/cs4780/2018fa/lectures/lecturenote19.html?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMML241ENSkillsNetwork31576874-2022-01-01">Kilian Weinberger</a>, we can write the average loss as:

$\ell(H) = \frac{1}{n}\sum_{i=1}^n \ell(H_T(\mathbf{x}_i), y_i)$

Here, $\ell$ is the loss for one data point, and $\ell(H)$ is the average loss over all data points. We assume the loss is convex, meaning it has a bowl shape like the figure below. We also assume it is differentiable, so its derivative exists.


```python
x = np.linspace(-1, 2)
fig, (ax1, ax2) = plt.subplots(1, 2)
ax1.plot(x, x**2, linewidth=2)
ax1.set_title('convex function')
ax2.plot(x, x**2 + np.exp(-5*(x - .5)**2), linewidth=2)
ax2.set_title('non-convex function')
plt.show()
```

At step $t$, Gradient Boosting already has a model from the previous steps, called $H_{t-1}(\mathbf{x})$. This is the current strong model. We keep this model fixed and train one new weak learner, called $h_t(\mathbf{x})$.

The updated model will be $H_{t-1}(\mathbf{x}) + \gamma h_t(\mathbf{x})$. The goal is to choose $h_t(\mathbf{x})$ so this updated model has a smaller loss. Gradient Boosting does not go back and retrain all the old weak learners. It only trains the next weak learner to improve the current model.

The formula below means: choose the weak learner $h_t(\mathbf{x})$ that minimizes the total loss on the training data.

$h_t = \operatorname*{argmin}_{h \in \mathbb{H}} \sum_{i=1}^n \ell(H_{t-1}(\mathbf{x}_i) + \gamma h(\mathbf{x}_i), y_i)$  (1)


The value $\gamma$ controls how much the new weak model changes the current model. Usually, $\gamma$ is selected using validation data. In some algorithms, such as AdaBoost, we can derive an expression for the optimal value of $\gamma$.


### **How to Minimize Cost**


At step $t$, we already have a model:

$H_{t-1}(\mathbf{x})$

This model makes predictions, but it still has errors. Gradient Boosting trains a new weak learner:

$h_t(\mathbf{x})$

to reduce those errors.

The new model becomes:

$H_t(\mathbf{x}) = H_{t-1}(\mathbf{x}) + \gamma h_t(\mathbf{x})$

Here, $\gamma$ is the learning rate. It controls how much the new weak learner changes the current model.


The question is:

**What should the next weak learner learn?**

Gradient Boosting answers this question by looking at the loss function. The loss tells us how wrong the current model is. We want the next weak learner to move the model in the direction that reduces the loss.

The gradient tells us the direction where the loss increases fastest. So the negative gradient tells us the direction where the loss decreases fastest.

That is why Gradient Boosting trains the next weak learner to fit the negative gradient.


For each training point, we compute:

$r_i = -\frac{\partial \ell(y_i, H_{t-1}(\mathbf{x}_i))}{\partial H_{t-1}(\mathbf{x}_i)}$

This value $r_i$ is called the pseudo-residual. It tells us what the next weak learner should try to predict for data point $i$.

The simple plot below shows the idea of a gradient. For a loss curve, the gradient is the slope at the current point. Gradient Boosting uses this slope information to find a direction that reduces the loss.


```python
plt.plot(x, x**2, linewidth=2, label='loss curve')
plt.plot(x, 2*x - 1, label='slope at one point')
plt.plot(1, 1, 'k+')
plt.title('Gradient as the slope of the loss curve')
plt.legend()
plt.show()
```

Then we train the next weak learner by solving:

$h_t = \operatorname*{argmin}_{h \in \mathbb{H}} \sum_{i=1}^{n} (r_i - h(\mathbf{x}_i))^2$

In simple words, we choose the weak learner that best predicts the pseudo-residuals.


The process is:

1. Use the current model to make predictions.
2. Compute the pseudo-residuals using the negative gradient of the loss.
3. Train the next weak learner to predict those pseudo-residuals.
4. Add the new weak learner to the current model.
5. Repeat.


The main idea is:

**Each new weak learner is trained to predict the direction that will reduce the current loss.**


### **Example with Python**


The math can feel abstract, so let's use a small Python example. We will use squared-error loss because it makes the main idea easy to see.

With squared-error loss, the pseudo-residual is the usual residual:

$\text{residual} = \text{actual value} - \text{current prediction}$

So each new weak learner tries to predict what the current model is still missing.


$\ell(H) = \sum_{i=1}^n (y_i - H_{t-1}(\mathbf{x}_i))^2$

$r_{i,t} = y_i - H_{t-1}(\mathbf{x}_i)$


Then we train the next weak learner to predict these residuals. In simple English, we choose the weak learner whose predictions are closest to the residuals.


$h_t = \operatorname*{argmin}_{h \in \mathbb{H}} \left\{ \sum_{i = 1}^{n} (r_{i,t} - h(\mathbf{x}_i))^2 \right\}$


Now let's walk through the idea in Python. We will create a toy dataset with one feature $x$. The target value has three flat levels:

* $y=0$ when $0 < x < 1$
* $y=1$ when $1 \leq x < 2$
* $y=2$ when $x \geq 2$

This simple shape helps us see how the boosted model improves step by step.


```python
X=np.linspace(0, 3,num=100)
 
y=np.zeros(X.shape)
y[X>1]=1
y[X>2]=2
plt.plot(X,y)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

In the first iteration, there is no previous model yet. So the first weak learner $h_1(\mathbf{x})$ tries to predict the original target values $y_i$.


$h_1 = \operatorname*{argmin}_{h \in \mathbb{H}} \left\{ \sum_{i = 1}^{n} (y_i - h(\mathbf{x}_i))^2 \right\}$


Here, we use a shallow decision tree. This tree is intentionally simple, so it is a weak learner.


```python
reg = DecisionTreeRegressor(max_depth=1)
reg.fit(X.reshape(-1,1),y)
h_1=reg.predict(X.reshape(-1,1))
```

Now we plot the first weak learner. It captures part of the pattern, but it does not match the target well by itself.


```python
plt.plot(X,y,label="y")
plt.plot(X,h_1,label="$h_{1}(x)$")
plt.legend()
plt.show()
```

For the second step, the current strong model is just the first weak learner:

$H_1(\mathbf{x}) = h_1(\mathbf{x})$

Then we calculate the residual. This residual is the part the first model did not explain.


$r_{i,2} = r_{i,1} - H_1(\mathbf{x}_i)$


```python
gamma=1
r=y-gamma*reg.predict(X.reshape(-1,1))
```

Next, we train the second weak learner $h_2(\mathbf{x})$ to predict this residual. This means the second learner is not trying to predict $y$ directly. It is trying to predict what the first learner missed.


$h_2 = \operatorname*{argmin}_{h \in \mathbb{H}} \left\{ \sum_{i = 1}^{n} (r_{i,2} - h(\mathbf{x}_i))^2 \right\}$


Then we add the second weak learner to the current model:

$H_2(\mathbf{x}) = H_1(\mathbf{x}) + \gamma h_2(\mathbf{x})$


We repeat the same process for each step $t$:

1. Fit a weak learner to the current residual.
2. Add the weak learner to the strong model.
3. Update the residual.


$r_{i,t} = r_{i,t-1} - \gamma h_{t-1}(\mathbf{x}_i)$


$h_t = \operatorname*{argmin}_{h \in \mathbb{H}} \left\{ \sum_{i = 1}^{n} (r_{i,t} - h(\mathbf{x}_i))^2 \right\}$


The strong model is updated by adding the new weak learner:

$H_t(\mathbf{x}) = H_{t-1}(\mathbf{x}) + \gamma h_t(\mathbf{x})$


We can perform these repeated updates with a Python loop. First, we write a function that makes a prediction by adding the predictions from all weak learners in the list `weak_learners`.


```python
def predict(y, weak_learners, gamma):
    yhat=np.zeros(y.shape)
    for h in weak_learners:
        yhat += gamma*h.predict(X.reshape(-1,1))
    return yhat

```

Then we train one weak learner at each iteration. We set $\gamma = 1$ in this simple example. After each step, we update the residual and plot the combined model. The plots are shown in rows, with four plots per row.


```python
weak_learners = []
gamma = 1
r = y.copy()
n_iterations = 10
n_cols = 4
n_rows = int(np.ceil(n_iterations / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 3.5*n_rows), sharex=True, sharey=True)
axes = axes.ravel()

for t_ in range(n_iterations):
    # Train a weak learner on the current residual.
    reg=DecisionTreeRegressor(max_depth=1)
    reg.fit(X.reshape(-1,1),r)
    weak_learners.append(reg)
    
    # Update the residual for the next iteration.
    r=r-gamma*reg.predict(X.reshape(-1,1))
    
    # Plot the combined model.
    ax = axes[t_]
    ax.plot(X, y, label="y")
    ax.plot(X, predict(y, weak_learners, gamma), label="$H_{}(x)$".format(t_+1))
    ax.set_title("Iteration {}".format(t_+1))
    ax.legend()

for ax in axes[n_iterations:]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
