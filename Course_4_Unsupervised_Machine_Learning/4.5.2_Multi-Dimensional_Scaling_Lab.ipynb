# **Multi-Dimensional Scaling**


## **Introduction**

Multi-Dimensional Scaling (**MDS**) is a dimensionality reduction method that represents high-dimensional data in a lower-dimensional space, usually 2D or 3D. Its main goal is to preserve the pairwise distances between data points as much as possible. If two points are close in the original data, they should stay close in the MDS embedding. If two points are far apart, they should remain far apart.



The image above compares three common dimensionality reduction methods: **MDS**, **t-SNE**, and **UMAP**. MDS focuses on preserving global pairwise distances. t-SNE focuses more on local neighborhoods, so it is often good at showing clusters. UMAP tries to preserve local structure and some global structure, and it is often faster on larger datasets.

In this lab, we focus on MDS. You will see how MDS takes distance or dissimilarity information and places objects in a lower-dimensional space for easier visualization and interpretation. Additional articles on MDS: <a href="https://link.springer.com/chapter/10.1007/978-3-642-82580-4_139?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMML0187ENSkillsNetwork31430127-2022-01-01">[1]</a>, <a href="https://www.djsresearch.co.uk/glossary/item/Multi-Dimensional-Scaling?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMML0187ENSkillsNetwork31430127-2022-01-01">[2]</a>


### **Use Cases of Multi-Dimensional Scaling**

MDS is useful when the main information we have is how similar or different objects are from each other. The introduction image shows this idea: MDS does not mainly search for variance like PCA. Instead, it tries to preserve distances in a lower-dimensional map.

Common use cases include:

* Recognizing families of parts in order to design cellular manufacturing systems.
* Creating groups of products when designing assembly areas.
* Market research, where MDS can plot product or brand perceptions in an easy-to-interpret visual space.
* Visualizing similarities between objects when the original feature space has too many dimensions to plot directly.

For instance, suppose a realtor has many listings to sell. Each listing has several attributes such as number of bedrooms, number of bathrooms, square feet, and price. As a data scientist, you are hired to find similarities and differences between the listings so brokers can make better recommendations for buyers.

Because the listings have more attributes than we can easily visualize, MDS can reproduce their relationships in a lower-dimensional space. Listings that are similar should appear close together, while listings that are different should appear farther apart.

There are several categories of Multidimensional Scaling (MDS). In this lab, we will review **Metric MDS** and **Non-Metric MDS** using the **scikit-learn** library. ## **Metric MDS**

Metric MDS is a dimensionality reduction method that tries to preserve the distances between data points when projecting high-dimensional data into a lower-dimensional space (usually 2D).

The main idea is simple:

> If two objects are close together in the original data, they should remain close together after projection.
>
> If two objects are far apart originally, they should also stay far apart in the new space.

### **Distance Matrix**

Suppose we have $N$ objects.

First, we calculate the distance between every pair of objects:

$$
d_{ij}
$$

where:

- $d_{ij}$ = distance between object $i$ and object $j$

These distances form a **distance matrix**:

$$
D =
\begin{bmatrix}
d_{11} & d_{12} & \cdots & d_{1N} \\
d_{21} & d_{22} & \cdots & d_{2N} \\
\vdots & \vdots & \ddots & \vdots \\
d_{N1} & d_{N2} & \cdots & d_{NN}
\end{bmatrix}
$$

This matrix contains the relative distances between all objects.

### **Goal of MDS**

MDS tries to place the objects into a lower-dimensional space:

$$
z_1, z_2, ..., z_N
$$

so that the new distances are as similar as possible to the original distances.

The new distance between two projected points is:

$$
\| z_i - z_j \|
$$

where:

- $z_i$ and $z_j$ are the projected coordinates
- $\|\cdot\|$ means Euclidean distance


### **Stress Function**

To measure how well the projection preserves the original distances, MDS uses a quantity called **Stress**.

$$
\text{Stress} =
\sqrt{
\sum_{i \neq j}
(d_{ij} - \|z_i - z_j\|)^2
}
$$

This equation measures the difference between:

- the original distance $d_{ij}$
- the projected distance $\|z_i - z_j\|$

The goal is:

- Small Stress → good projection
- Large Stress → poor projection

MDS repeatedly adjusts the positions of the projected points to minimize Stress.


### **Simple Example**

Imagine you know the distances between several cities, but you do not know their exact coordinates.

For example:

| City Pair | Distance |
|---|---|
| Boston → New York | 306 km |
| Boston → Chicago | 1360 km |
| New York → Chicago | 1145 km |

MDS can reconstruct approximate city positions using only these distances.

So MDS converts:

> Relative distances → approximate spatial locations


MDS focuses on preserving pairwise distances.

Unlike PCA, MDS does not directly analyze features or variance. Instead, it only cares about how similar or far apart the objects are. This makes MDS useful when distance or similarity information is more important than the original feature values.

### **Example 1**

In this case, we can recreate the distance dataframe using `pdist`. This calculates the Pairwise distances between data points in df. We then apply `squareform` that converts the result to a square-form distance matrix:

$\sqrt{(lon_i-lon_j)^2 + (lat_i-lat_j)^2 }$


````python
from scipy.spatial.distance import squareform, pdist

distance = pd.DataFrame(
    squareform(pdist(df.iloc[:, 1:])), columns=df.index, index=df.index
)
````

We see MDS clusters on the distance matrix; unlike the principal component analysis, we can apply different distances. In the following  lines of code, we use other distance metrics on the longitude and latitude of different cities, then apply MDS. We plot out the results and see each distance has its own unique property.


````python
dist = [
    "cosine",
    "cityblock",
    "seuclidean",
    "sqeuclidean",
    "cosine",
    "hamming",
    "jaccard",
    "chebyshev",
    "canberra",
    "braycurtis",
]
plot_points(df, title="original dataset")
for d in dist:

    distance = pd.DataFrame(
        squareform(pdist(df.iloc[:, 1:], metric=d)),
        columns=df.index,
        index=df.index,
    )

    embedding = MDS(dissimilarity="precomputed", random_state=0, n_components=2)
    X_transformed = embedding.fit_transform(distance)
    df_t = pd.DataFrame(X_transformed, columns=df.columns, index=df.index)

    plot_points(df_t, title="Embedded Coordinates using " + d, color="blue")
````

## **Non-Metric MDS**


In  Non-Metric MDS, we apply a function $f(.)$, a distance metric, $d(z_i,z_j)$, for the distances between the points in the embedding space.  $f(.)$ is a monotonic function. This means it's always increasing, this preserves the order of the distance. For example, if one distance is larger before applying the function, it will still be larger after applying the function.


$\text{Stress}*D(z\_1,z\_2,...,z_N)=\Biggl(\frac{\sum*{i\ne j=1,...,N}\bigl(f(d(z_i,z_j))-d\_{i,j}\bigr)^2}{\sum d\_{i,j}^2}\Biggl)^{1/2}$


To  perform non-metric MDS , we first set `metric` to False.


````python
metric = False
embedding = MDS(
    dissimilarity="precomputed", n_components=2, metric=metric, random_state=0
)
````

We then apply `fit_transform()` to the data from array distance that fits the model and returns the embedded coordinates, and convert it to a dataframe:


````python
X_transformed = embedding.fit_transform(distance)
df_t = pd.DataFrame(X_transformed, columns=df.columns, index=df.index)
df_t.head(8)
````

Non-metric MDS does better at preserving the horizontal scale, but the cities are negatively scaled in the longitudinal direction. At the same time, some cities like Paris and Montreal seem to have a different relationship  in the latitudinal direction.


````python
plot_points(df, title="original dataset")

plot_points(
    df_t, color="blue", title="Embedded Coordinates using Euclidean distance "
)
````

We can also apply different distances to Non-Metric MDS.


````python
dist = [
    "cosine",
    "cityblock",
    "seuclidean",
    "sqeuclidean",
    "cosine",
    "hamming",
    "jaccard",
    "chebyshev",
    "canberra",
    "braycurtis",
]
plot_points(df, title="original dataset")
metric = True
for d in dist:

    distance = squareform(pdist(df.iloc[:, 1:], metric=d))
    distance = np.nan_to_num(distance, nan=0.0, posinf=0.0, neginf=0.0)
    np.fill_diagonal(distance, 0.0)

    embedding = MDS(
        dissimilarity="precomputed",
        random_state=0,
        n_components=2,
        metric=metric,
        n_init=1,
    )
    X_transformed = embedding.fit_transform(distance)
    df_t = pd.DataFrame(X_transformed, columns=df.columns, index=df.index)

    plot_points(df_t, title="Embedded Coordinates using " + d, color="blue")
````

## **Dimensionality Reduction with MDS**


MDS can also be used for Dimensionality Reduction. We just need to set $d\_{ij}=||x_i-x_j||\ $, where the dimension of $x_i$ is greater than $z_j$. Consider the  digits dataset that consists of 8x8 pixel images of digits ranging from 0 to 9.


````python
from sklearn.datasets import load_digits

digits = load_digits(n_class=6)
X, y = digits.data, digits.target
n_samples, n_features = X.shape
````

We have 64 features or dimensions of $x_i$


````python
print("samples:", n_samples, "features", n_features)
````

We can plot out the different $x_i$  by converting them to a square array:


````python
fig, axs = plt.subplots(nrows=10, ncols=10, figsize=(6, 6))
for idx, ax in enumerate(axs.ravel()):
    ax.imshow(X[idx].reshape((8, 8)), cmap=plt.cm.binary)
    ax.axis("off")
_ = fig.suptitle(
    "A selection from the 64-dimensional digits dataset", fontsize=16
)
````

We create an embedding object. As we are using the features X,  we set dissimilarity to `euclidean`, which is the default choice for the `dissimilarity` parameter in **sklearn.manifold.MDS**.


````python
embedding = MDS(n_components=2, n_init=1, max_iter=120, n_jobs=2)
````

We can apply MDS to the data X, we will fit the model and make a prediction simultaneously using the <code>fit_transform</code> method.


````python
X_transformed = embedding.fit_transform(X)
````

We plot the samples in the embedding space and overlay images of the digit and the colour representing the digit. We see that although there is some overlap, each cluster corresponds to a different digit, this is despite the fact that the method is unsupervised.


````python
fig, ax = plt.subplots()
plot_embedding(X_transformed, "Metric MDS ", ax)
plt.show()
````
